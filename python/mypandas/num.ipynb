```python
import pandas as pd
coffee=pd.read_csv(r"C:\Users\Prakhar\Downloads\coffee.csv")
# coffee["price"]=4.99

# to be more specific
import numpy as np
coffee["price"]=np.where(coffee["Coffee Type"]=="Latte",3.99,6)
coffee.head()
```

```python
# to remove a row or a column 
# coffee.drop(0)
coffee.drop(columns=["price"])
# to rename a column
coffee.rename(columns={'price':'new_price'})
```
